```python
from transformers import pipeline
import torch
```

```python
pipe = pipeline(
    "text-generation", 
    model="google/gemma-3-4b-it", 
    torch_dtype="auto")
```

```python
# ZERO-SHOT PROMPTING

messages = [
    {
        "role": "system",
        "content": [{"type": "text", "text": "You are a text generator."}]
    },
    {
        "role": "user",
        "content": [
            {"type": "text", "text": f"Generate 200 new idioms and slang terms.\n"
             f"Only respond with the results of your slang and idiom generation. Don't generate any additional text.\n" 
             f"The results should contain an equal amount of slang and idioms with a total of 200 new idioms and slang terms. Do not provide only slang or only idioms.\n"
             f"Each generated slang/idiom should be separated with a colon. Once again, don't generate any additional text like **Slang (100 words)**.\n"
             f"The newly-generated idioms/slang should be novel and semantically fluent. Idioms are phrases longer than 2-3 words.\n" 
             f"Here is an example of the format of the results: \n"
             f"yeet: spill the tea: flex: throw shade: rizzler: passed with flying colors: crispy: joe mama\n"}
        ]
    }
]

output = pipe(
    messages, 
    max_new_tokens=2000, 
    temperature=0.7, 
    top_p=0.9)
```

```python
print(output[0]["generated_text"][-1]["content"])
```

```python
results = output[0]["generated_text"][-1]["content"].split(": ")
for i, slang in enumerate(results):
        print(f"{i+1}. {slang.strip()}")
```

```python
import json

with open("generated_slang.json", "w") as f:
    json.dump(results, f, indent=4)
```

```python
# FEW-SHOT PROMPTING

messages = [
    {
        "role": "system",
        "content": [{"type": "text", "text": "You are a text generator."}]
    },
    {
        "role": "user",
        "content": [
            {"type": "text", "text": f"Generate 200 new idioms and slang terms.\n"
             f"Only respond with the results of your slang and idiom generation. Don't generate any additional text.\n" 
             f"The results should contain an equal amount of slang and idioms with a total of 200 new idioms and slang terms. Do not provide only slang or only idioms.\n"
             f"Each generated slang/idiom should be separated with a colon. Once again, don't generate any additional text like **Slang (100 words)**.\n"
             f"The newly-generated idioms/slang should be novel and semantically fluent. Idioms are phrases longer than 2-3 words.\n" 
             f"Here is an example of the format of the results: \n"
             f"yeet: spill the tea: flex: throw shade: rizzler: passed with flying colors: crispy: joe mama\n"
             f"Below are a few examples of novel and semantically fluent slang/idioms and novel but not semantically fluent slang/idioms to help you generate new slang/idioms that are novel and semantically fluent:\n"
             f"Example 1: 'yeet' - not novel, semantically fluent\n"
             f"Example 2: 'glimmerflux' - novel, semantically fluent\n"
             f"Example 3: 'sad aioj ads' - novel, not semantically fluent\n"
             f"Example 4: 'flibber flabber' - novel, semantically fluent\n"
             f"Example 5: 'omega alpha sigma beta' - novel, not semantically fluent\n"
             f"Example 6: 'I alone am the chosen one' - not novel, semantically fluent\n"}        
        ]
    }
]

output_few = pipe(
    messages, 
    max_new_tokens=2000, 
    temperature=0.7, 
    top_p=0.9)
```

```python
print(output_few[0]["generated_text"][-1]["content"])
```

```python
results = output_few[0]["generated_text"][-1]["content"].split(": ")
for i, slang in enumerate(results):
        print(f"{i+1}. {slang.strip()}")
```

```python
with open("generated_slang_few_prompt.json", "w") as f:
    json.dump(results, f, indent=4)
```

```python
from transformers import pipeline
import torch
```

```python
pipe = pipeline(
    "text-generation", 
    model="google/gemma-3-4b-it", 
    torch_dtype="auto")
```

```python
messages = [
    {
        "role": "system",
        "content": [{"type": "text", "text": "You are a helpful assistant."}]
    },
    {
        "role": "user",
        "content": [
            {"type": "text", "text": "Generate 200 new slang. The newly-generated slang can be singular words or phrases. Only respond with the results of your slang generation. Each slang should be separated with a comma. The results should contain 200 slang in total."}
        ]
    }
]

output = pipe(
    messages, 
    max_new_tokens=2000, 
    temperature=0.7, 
    top_p=0.9)
```

```python
print(output[0]["generated_text"][-1]["content"])
```

```python
results = output[0]["generated_text"][-1]["content"].split(",")
for i, slang in enumerate(results):
    print(f"{i+1}. {slang.strip()}")
```

```python
import json

with open("generated_slang.json", "w") as f:
    json.dump(results, f, indent=4)
```
